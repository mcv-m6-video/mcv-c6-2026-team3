# Video to GIF Converter

This notebook takes detection and mask videos from a results directory and creates GIFs for a specific time segment.

```python
import os
from moviepy import VideoFileClip

# ============= CONFIGURATION =============
RESULTS_DIR = "../results/adaptive_gaussian_run"
#RESULTS_DIR = "../results/gaussian_shadow_run"
#RESULTS_DIR = "../results/mog2_run"
START_TIME = 37
END_TIME = 41
OUTPUT_NAME_PREFIX = "mog2_sample"  # Prefix for output GIF names
# =========================================

print(f"Configuration:")
print(f"  Directory: {RESULTS_DIR}")
print(f"  Time range: {START_TIME}s - {END_TIME}s")
print(f"  Output prefix: {OUTPUT_NAME_PREFIX}")
```

```python
# Generate GIF from detections video
OUTPUT_DETECTIONS_GIF = f"{OUTPUT_NAME_PREFIX}_detections.gif"

try:
    detections_video_path = os.path.join(RESULTS_DIR, "detections.avi")
    output_detections_path = os.path.join(RESULTS_DIR, OUTPUT_DETECTIONS_GIF)
    
    if not os.path.exists(detections_video_path):
        print(f"⚠️  Detections video not found at {detections_video_path}")
    else:
        print(f"Processing detections video...")
        video = VideoFileClip(detections_video_path)
        clip = video.subclipped(START_TIME, END_TIME)
        clip.write_gif(output_detections_path, fps=20)
        clip.close()
        video.close()
        print(f"✓ Detections GIF saved at: {output_detections_path}")
        
except Exception as e:
    print(f"✗ Error generating detections GIF: {e}")
```

```python
# Generate GIF from mask video
OUTPUT_MASK_GIF = f"{OUTPUT_NAME_PREFIX}_mask.gif"

try:
    mask_video_path = os.path.join(RESULTS_DIR, "mask.avi")
    output_mask_path = os.path.join(RESULTS_DIR, OUTPUT_MASK_GIF)
    
    if not os.path.exists(mask_video_path):
        print(f"⚠️  Mask video not found at {mask_video_path}")
    else:
        print(f"Processing mask video...")
        mask_video = VideoFileClip(mask_video_path)
        mask_clip = mask_video.subclipped(START_TIME, END_TIME)
        mask_clip.write_gif(output_mask_path, fps=10)
        mask_clip.close()
        mask_video.close()
        print(f"✓ Mask GIF saved at: {output_mask_path}")
        
except Exception as e:
    print(f"✗ Error generating mask GIF: {e}")
```
