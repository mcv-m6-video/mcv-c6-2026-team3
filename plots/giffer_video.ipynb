# Video to GIF Converter

This notebook takes a detection video from `results/adaptive_gaussian_simple_params_run` and creates a GIF for a specific time segment (33s - 40s).

```python
import os
from moviepy import VideoFileClip

# Configuration
RESULTS_DIR = "../results/adaptive_gaussian_run"
OUTPUT_GIF = "adaptive_gaussian_sample.gif"

# Time segment in seconds
START_TIME = 37
END_TIME = 41

# Find the video file (assuming .avi, prioritize detections.avi)
video_files = [f for f in os.listdir(RESULTS_DIR) if f.endswith('.avi')]

if not video_files:
    raise FileNotFoundError(f"No .avi files found in {RESULTS_DIR}")

# Prefer 'detections.avi' if present, otherwise 'mask.avi', otherwise take the first one
if "detections.avi" in video_files:
    target_video = "detections.avi"
elif "mask.avi" in video_files:
    target_video = "mask.avi"
else:
    target_video = video_files[0] 

print(f"Processing video: {target_video}")

video_path = os.path.join(RESULTS_DIR, target_video)
output_path = os.path.join(RESULTS_DIR, OUTPUT_GIF)
```

```python
# Generate GIF
try:
    video = VideoFileClip(video_path)
    
    # Cut the clip
    clip = video.subclipped(START_TIME, END_TIME)
    
    # Resize if necessary to keep file size reasonable (optional)
    # clip = clip.resized(width=640) 
    
    # Write GIF
    clip.write_gif(output_path, fps=10)  # fps=10 is usually enough for gifs
    
    # Close the clips to free resources
    clip.close()
    video.close()
    
    print(f"GIF saved successfully at: {output_path}")
except Exception as e:
    print(f"Error generating GIF: {e}")
```

```python
# Generate GIF from mask video
OUTPUT_MASK_GIF = "adaptive_gaussian_mask_sample.gif"

try:
    mask_video_path = os.path.join(RESULTS_DIR, "mask.avi")
    output_mask_path = os.path.join(RESULTS_DIR, OUTPUT_MASK_GIF)
    
    if not os.path.exists(mask_video_path):
        raise FileNotFoundError(f"Mask video not found at {mask_video_path}")
    
    mask_video = VideoFileClip(mask_video_path)
    
    # Cut the clip
    mask_clip = mask_video.subclipped(START_TIME, END_TIME)
    
    # Resize if necessary to keep file size reasonable (optional)
    # mask_clip = mask_clip.resized(width=640) 
    
    # Write GIF
    mask_clip.write_gif(output_mask_path, fps=10)
    
    # Close the clips to free resources
    mask_clip.close()
    mask_video.close()
    
    print(f"Mask GIF saved successfully at: {output_mask_path}")
except Exception as e:
    print(f"Error generating mask GIF: {e}")
```
