# Video to GIF Converter

This notebook takes detection and mask videos from a results directory and creates GIFs for a specific time segment.

**Features:**
- Extract a specific time segment from videos
- Configurable output FPS
- **Speed multiplier**: Control playback speed (x1 = normal, x2 = double speed, etc.)
  - Faster speeds result in smaller file sizes
  - Same frames are used, just played back faster

```python
import os
from moviepy.editor import VideoFileClip

# ============= CONFIGURATION =============
RESULTS_DIR = "../results/tracking_iou_finetuned"
RESULTS_DIR= "../results"
START_TIME = 60+57
END_TIME = 120+2
OUTPUT_NAME_PREFIX = "tolo_sample2"  # Prefix for output GIF names

# Speed multiplier: x1 = normal speed, x2 = double speed (faster), x0.5 = half speed (slower)
# Higher values = faster GIF = smaller file size (fewer frames total)
SPEED_MULTIPLIER = 1

# GIF FPS (frames per second) - Lower = fewer frames = smaller file
# Typical values: 24 (smooth), 15 (balanced), 10 (small), 5 (very small)
GIF_FPS = 10

# Resolution scale (optional) - Lower = smaller file
# 1.0 = original, 0.5 = half size, 0.3 = tiny
RESOLUTION_SCALE = 0.75
# =========================================

print(f"Configuration:")
print(f"  Directory: {RESULTS_DIR}")
print(f"  Time range: {START_TIME}s - {END_TIME}s")
print(f"  Output prefix: {OUTPUT_NAME_PREFIX}")
print(f"  Speed multiplier: x{SPEED_MULTIPLIER}")
print(f"  GIF FPS: {GIF_FPS}")
print(f"  Resolution scale: {RESOLUTION_SCALE}")
```

```python
# Generate GIF from detections video
OUTPUT_DETECTIONS_GIF = f"{OUTPUT_NAME_PREFIX}_tracking.gif"

try:
    detections_video_path = os.path.join(RESULTS_DIR, "tracking.avi")
    output_detections_path = os.path.join(RESULTS_DIR, OUTPUT_DETECTIONS_GIF)
    
    if not os.path.exists(detections_video_path):
        print(f"⚠️  Detections video not found at {detections_video_path}")
    else:
        print(f"Processing detections video...")
        video = VideoFileClip(detections_video_path)
        clip = video.subclip(START_TIME, END_TIME)
        
        # Apply speed multiplier if different from 1.0
        if SPEED_MULTIPLIER != 1.0:
            clip = clip.speedx(SPEED_MULTIPLIER)
        
        # Apply resolution scaling if different from 1.0
        if RESOLUTION_SCALE != 1.0:
            clip = clip.resize(RESOLUTION_SCALE)
        
        # Force FPS reduction to actually reduce file size
        # This tells moviepy to sample the video at the specified FPS
        clip = clip.set_fps(GIF_FPS)
        
        # Write GIF with controlled FPS
        clip.write_gif(output_detections_path, fps=GIF_FPS)
        
        clip.close()
        video.close()
        print(f"✓ Detections GIF saved at: {output_detections_path}")
        print(f"  Speed: x{SPEED_MULTIPLIER}, FPS: {GIF_FPS}, Resolution: {RESOLUTION_SCALE}")
        
        # Show file size
        file_size_mb = os.path.getsize(output_detections_path) / (1024 * 1024)
        print(f"  File size: {file_size_mb:.2f} MB")
        
except Exception as e:
    print(f"✗ Error generating detections GIF: {e}")
```
